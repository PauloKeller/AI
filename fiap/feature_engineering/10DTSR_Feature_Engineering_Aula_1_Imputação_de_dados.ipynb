# Exercício

Realizar o Data Inputation utilizando o dataset abaixo para os seguintes casos:

- Valor arbitrário: -100.000.000
- Mediana
- Média
- End of Tail (em +3sigma)
- KNN Inputer

**Dica**: Cuidado para não utilizar o mesmo dataset nos diferentes exercícios

```python
# knn imputation transform for the horse colic dataset
from numpy import isnan
import pandas as pd

# Carrega  dataset
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/horse-colic.csv'
df = pd.read_csv(url, header=None, na_values='?')

df
```

```python
#Ajuste Para separar a coluna target

target_column = 23
y = df[target_column]
x_columns = df.columns.to_list()
x_columns.remove(target_column)
X = df[x_columns]
```

```python
#Total missing
X.isnull().sum()
```

## Respostas abaixo:

```python
X
```

```python
arbitrary_imputer = SimpleImputer(strategy='constant', fill_value=-100_000_000)
X_arbitrary = pd.DataFrame(arbitrary_imputer.fit_transform(X), columns=X.columns)

X_arbitrary
```

```python
median_imputer = SimpleImputer(strategy='median')
X_median = pd.DataFrame(median_imputer.fit_transform(X), columns=X.columns)

X_median
```

```python
mean_imputer = SimpleImputer(strategy='mean')
X_mean = pd.DataFrame(mean_imputer.fit_transform(X), columns=X.columns)

X_mean
```

```python
def end_of_tail_impute(df):
    df_imputed = df.copy()
    for col in df.columns:
        mean = df[col].mean()
        std = df[col].std()
        end_tail_value = mean + 3 * std
        df_imputed[col].fillna(end_tail_value, inplace=True)
    return df_imputed
```

```python
X_end_tail = end_of_tail_impute(X)

X_end_tail
```

```python
knn_imputer = KNNImputer(n_neighbors=2)
X_knn = pd.DataFrame(knn_imputer.fit_transform(X), columns=X.columns)

X_knn
```
