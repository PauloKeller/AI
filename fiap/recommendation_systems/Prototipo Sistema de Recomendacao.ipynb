```python
import pandas as pd
import json
import openpyxl
```

# 1. Base de dados

## 1.1 Importando e tratando a base de clientes e produtos

```python
# Transformar a base "clientes.json" em um DataFrame do Pandas chamado clientes_df, O ID deve ser o índice do DataFrame

# Lê o arquivo JSON
with open('clientes.json', 'r') as file:
    data = json.load(file)

# Cria o DataFrame a partir dos dados JSON
clientes_df = pd.DataFrame(data)

# Define a coluna 'id' como índice do DataFrame
clientes_df.set_index('id', inplace=True)

# Exibe o DataFrame resultante
display(clientes_df)
```

## 1.2 Verificandos como está o DataFrame

```python
clientes_df.info()
```

```python
# Converte as colunas Conta Corrente e Poupança para float
clientes_df['Conta Corrente'] = clientes_df['Conta Corrente'].astype(float)
clientes_df['Poupanca'] = clientes_df['Poupanca'].astype(float)

clientes_df.info()
```

## 1.3 Substituindo os valores nulos por 0

```python
# Preenche valores ausentes com 0
clientes_df.fillna(0, inplace=True)
clientes_df.info()
```

# 2. Recomendação de produtos

## 2.1 Calculando distancia euclidiana entre os clientes

```python
# Verifica a semelhança entre os clientes com base na Conta Corrente, poupança e idade (Vetor 3D)
# Calcula a distancia euclidiana entre os clientes e cria uma coluna no final do dataframe informando o ID do cliente mais próximo
# Utiliza a diferença entre as idades e o saldo da conta corrente e poupança para calcular a distância euclidiana


def euclidean_distance(row1, row2):
    # Diferença ao quadrado entre as idades, Conta Corrente e Poupança
    age_diff = (row1['idade'] - row2['idade']) ** 2
    cc_diff = (row1['Conta Corrente'] - row2['Conta Corrente']) ** 2
    cp_diff = (row1['Poupanca'] - row2['Poupanca']) ** 2
    return (age_diff + cc_diff + cp_diff) ** 0.5

# Cria uma coluna para armazenar o ID do cliente mais próximo
clientes_df['id_cliente_mais_proximo'] = None

# Calcula a distância euclidiana entre cada cliente e todos os outros clientes
for index, row in clientes_df.iterrows():
    distances = []
    for index2, row2 in clientes_df.iterrows():
        if index != index2:  # Não comparar o cliente consigo mesmo
            dist = euclidean_distance(row, row2)
            distances.append((index2, dist))
    
    # Encontra o ID do cliente mais próximo
    closest_client = min(distances, key=lambda x: x[1])
    clientes_df.at[index, 'id_cliente_mais_proximo'] = closest_client[0]

# Exibe o DataFrame atualizado com a coluna do cliente mais próximo
display(clientes_df.head(5))
```

## 2.2 Criar coluna com lista dos serviços contratados por cada cliente




```python
# Cria uma nova nova coluna com a lista de produtos contratados por cada cliente, a partir das colunas: Conta Corrente, 
# Poupanca, Credito Imobiliario, Fundos de Investimento de Acoes, Cartao de Credito, PGBL, ETF, LCI, 
# Fundos de Investimento Multimercado, Consorcio, Fundo de Capitalizacao, LCA, Fundos de Investimento de Renda Fixa, 
# VGBL, Credito Pessoal, FII, CDB, Acoes. Quando o valor for maior que 0, o produtos deve ser adicionado à lista, 
# caso contrário não deve ser adicionado.   

def get_services(row):
    services = []
    if row['Conta Corrente'] > 0:
        services.append('Conta Corrente')
    if row['Poupanca'] > 0:
        services.append('Poupanca')
    if row['Credito Imobiliario'] > 0:
        services.append('Credito Imobiliario')
    if row['Fundos de Investimento de Acoes'] > 0:
        services.append('Fundos de Investimento de Acoes')
    if row['Cartao de Credito'] > 0:
        services.append('Cartao de Credito')
    if row['PGBL'] > 0:
        services.append('PGBL')
    if row['ETF'] > 0:
        services.append('ETF')
    if row['LCI'] > 0:
        services.append('LCI')
    if row['Fundos de Investimento Multimercado'] > 0:
        services.append('Fundos de Investimento Multimercado')
    if row['Consorcio'] > 0:
        services.append('Consorcio')
    if row['Fundo de Capitalizacao'] > 0:
        services.append('Fundo de Capitalizacao')
    if row['LCA'] > 0:
        services.append('LCA')
    if row['Fundos de Investimento de Renda Fixa'] > 0:
        services.append('Fundos de Investimento de Renda Fixa')
    if row['VGBL'] > 0:
        services.append('VGBL')
    if row['Credito Pessoal'] > 0:
        services.append('Credito Pessoal')
    if row['FII'] > 0:
        services.append('FII')
    if row['CDB'] > 0:
        services.append('CDB')
    if row['Acoes'] > 0:
        services.append('Acoes')

    return ', '.join(services)

# Aplica a função para criar a nova coluna de serviços
clientes_df['produtos_contratados'] = clientes_df.apply(get_services, axis=1)

# Exibe o DataFrame atualizado com a nova coluna de serviços contratados
display(clientes_df.head(5))
```

## 2.3 Recomendando produtos financeiros para os clientes


```python
# Cria uma nova coluna chamada produtos recomendados, que contem os produtos recomendados para cada cliente, 
# com base nos produtos contratados e os produtos do cliente mais próximo (Diferença).

def recommend_products(row):
    recommended_products = set()
    actual_products = set()
    
    # Produtos contratados pelo cliente
    if row['produtos_contratados']:
        actual_products = row['produtos_contratados'].split(', ')
    
    # Adiciona os produtos do cliente mais próximo
    closest_client_id = row['id_cliente_mais_proximo']

    if closest_client_id in clientes_df.index:
        closest_client_products = clientes_df.at[closest_client_id, 'produtos_contratados'].split(', ')

        if closest_client_products:
            # Recomenda os produtos do cliente mais próximo que o cliente atual não possui
            for product in closest_client_products:
               if product not in actual_products:
                    recommended_products.add(product.strip())

    return ', '.join(recommended_products)

# Aplica a função para criar a nova coluna de produtos recomendados
clientes_df['produtos_recomendados'] = clientes_df.apply(recommend_products, axis=1)

# Exibe o DataFrame atualizado com a nova coluna de produtos recomendados
display(clientes_df.head(5))

# Salvar o DataFrame atualizado em um arquivo Excel chamado "clientes.xlsx"
clientes_df.to_excel('Produtos_Recomendados.xlsx', engine='openpyxl')
```
